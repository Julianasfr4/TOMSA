## 1. Análise crítica dos objetivos definidos *versus* objetivos atingidos


### 1.1. Enquadramento e objetivos inicialmente definidos

Na primeira entrega do trabalho de Tópicos de Matemática para Sistemas Autónomos (TOMSA), foi proposta a implementação de uma aplicação web interativa destinada à visualização e análise de missões de veículos robóticos marinhos, utilizando dados disponibilizados pelo INESC TEC, provenientes de missões realizadas na Aguçadoura.

O problema identificado consistia na dificuldade de interpretar diretamente os registos de telemetria dos robôs, uma vez que estes contêm informação temporal, coordenadas geográficas, orientação e velocidades, distribuída por diferentes ficheiros CSV.

O principal objetivo consistia, assim, em desenvolver uma ferramenta capaz de processar estes dados e permitir a análise do movimento dos veículos, tanto num referencial global como num referencial local associado a um robô selecionado.

Para concretizar esta solução, foram definidos cinco requisitos principais:

1. **Global View:** representação das posições e trajetórias dos três robôs num mapa geográfico.
2. **Local View:** representação dos restantes robôs relativamente à posição e orientação de um robô de referência.
3. **Mission Replay:** reprodução temporal da missão através de um slider e controlos de reprodução.
4. **Configuração YAML:** parametrização da aplicação através de um ficheiro de configuração.
5. **Output:** geração e exportação de dados processados a partir da aplicação.

Adicionalmente, foi proposta a integração de funcionalidades complementares de análise de proximidade, deteção de anomalias e interpretação da missão com recurso a inteligência artificial.

### 1.2. Análise dos objetivos atingidos

Após o desenvolvimento do projeto, foi possível implementar os cinco requisitos principais definidos na primeira entrega, bem como diversas funcionalidades complementares.

A **Global View** foi desenvolvida através da integração das bibliotecas Dash e Plotly, permitindo representar as trajetórias dos três robôs num mapa geográfico. Cada veículo é identificado por uma cor distinta, sendo a respetiva posição atualizada durante a reprodução da missão. A aplicação utiliza mapas OpenStreetMap, complementados por informação marítima OpenSeaMap, quando existe ligação à internet.

Relativamente à **Local View**, foi implementada uma representação espacial centrada no robô selecionado pelo utilizador. As posições dos restantes veículos são transformadas do referencial global NED para o referencial local do robô de referência, considerando a sua orientação. A aplicação disponibiliza representações bidimensionais e tridimensionais, incluindo a possibilidade de utilizar apenas o *heading* ou a rotação tridimensional completa.

O requisito **Mission Replay** foi concretizado através de um sistema de reprodução temporal que permite iniciar, interromper e controlar a evolução da missão. O utilizador pode selecionar diretamente um instante através do slider ou introduzir um valor temporal específico, sendo também possível alterar a velocidade de reprodução.

A **configuração YAML** foi implementada de forma a permitir a definição dos veículos, ficheiros de entrada, cores, referência geográfica, parâmetros de sincronização e critérios de análise. Adicionalmente, foi utilizada a biblioteca Pydantic para validar a estrutura e os valores da configuração.

Quanto ao requisito **Output**, foram desenvolvidas funcionalidades de exportação de telemetria processada em CSV, relatórios de missão em formato TXT e configurações em YAML. Os parâmetros de exportação podem ser definidos no ficheiro de configuração, incluindo os campos pretendidos, o referencial e o intervalo temporal.

Para além destes requisitos, foram implementadas operações de transformação de coordenadas geográficas para os referenciais ECEF e NED, conversão de quaternions para representações de orientação e sincronização dos dados dos diferentes robôs através de uma grelha temporal comum.

A interpolação das posições é realizada de forma linear, enquanto a orientação utiliza interpolação esférica de quaternions (*SLERP*), permitindo tratar adequadamente a evolução temporal das rotações.

Foram ainda desenvolvidos módulos de análise que permitem calcular distâncias entre veículos, identificar situações de proximidade e estimar o instante e a distância de maior aproximação (*TCPA* e *DCPA*), considerando um modelo de velocidade constante.

A aplicação inclui igualmente mecanismos de deteção de anomalias nos dados, nomeadamente saltos de posição e falhas temporais de telemetria.

### 1.3. Comparação entre a metodologia proposta e a implementação realizada

A arquitetura final manteve os princípios fundamentais definidos na primeira entrega, nomeadamente a separação entre processamento de dados, operações matemáticas, análise e visualização.

O projeto encontra-se organizado em módulos independentes, distribuídos por diferentes diretórios, como `config`, `geo`, `io_dados`, `analysis`, `ui`, `ai` e `tests`.

Esta organização permite separar as responsabilidades de cada componente e facilita a manutenção e evolução da aplicação.

Uma das decisões mais relevantes consistiu em realizar o pré-processamento dos dados antes de iniciar a interface. Desta forma, as operações de conversão de coordenadas, sincronização temporal e preparação dos indicadores da missão são executadas previamente, evitando repetir cálculos desnecessários durante a reprodução.

Outro aspeto importante foi a implementação própria das transformações geodésicas com recurso ao NumPy, utilizando as bibliotecas `pymap3d` e `pyproj` como referências de comparação nos testes. Esta abordagem permite relacionar diretamente os conceitos matemáticos estudados na unidade curricular com a implementação computacional.

Relativamente à componente de inteligência artificial, a solução final apresenta algumas diferenças face à proposta inicial. Em vez de depender exclusivamente de um serviço externo, foi implementada a possibilidade de utilizar um modelo de linguagem local através do Ollama, mantendo também suporte para uma API externa.

Adicionalmente, foi desenvolvido um mecanismo alternativo de geração de explicações baseado em regras, permitindo produzir informação interpretativa mesmo quando o modelo de linguagem não está disponível.

Embora a proposta inicial previsse a utilização de *background callbacks* para as operações de inteligência artificial, a implementação final utiliza um callback específico para solicitar a explicação. Esta diferença representa uma alteração à estratégia inicialmente definida, mantendo-se a separação lógica entre a reprodução temporal e a análise assistida por IA.

### 1.4. Limitações e aspetos a melhorar

Apesar da implementação dos requisitos principais, existem aspetos que devem ser considerados na avaliação crítica da solução.

A visualização geográfica depende da disponibilidade de serviços externos para apresentar os mapas de fundo. Embora tenha sido implementado um modo alternativo sem ligação à internet, este não disponibiliza o mesmo nível de informação geográfica.

A interpretação das orientações depende também da convenção de referenciais adotada. A configuração permite especificar a convenção utilizada, sendo importante garantir a sua coerência com os dados originais.

Relativamente à sincronização temporal, a aplicação limita a interpolação em intervalos com falhas superiores ao limite configurado. Esta decisão evita representar posições estimadas em períodos sem informação suficiente, embora possa originar descontinuidades na visualização.

A análise de proximidade através de DCPA e TCPA utiliza um modelo bidimensional de velocidade constante, no qual a velocidade horizontal é aproximada a partir da velocidade escalar e do *heading*. Esta aproximação pode ser menos adequada quando a direção do deslocamento não coincide com a orientação do veículo, pelo que os valores obtidos são indicadores aproximados e não previsões exatas das trajetórias futuras.

Por fim, a componente de inteligência artificial depende da disponibilidade e configuração do modelo selecionado. O mecanismo alternativo baseado em regras permite manter a funcionalidade de explicação, mas apresenta menor flexibilidade na interpretação dos acontecimentos.

### 1.5. Balanço crítico

A comparação entre os objetivos inicialmente definidos e a implementação final permite concluir que o projeto atingiu, ao nível das funcionalidades desenvolvidas, os cinco requisitos principais estabelecidos na primeira entrega.

Para além da reprodução visual das trajetórias, a aplicação integra ferramentas de processamento, transformação e análise dos dados, permitindo explorar o comportamento individual e relativo dos três robôs ao longo da missão.

O desenvolvimento do projeto constituiu uma oportunidade para aplicar conceitos fundamentais de TOMSA, particularmente transformações entre referenciais, matrizes de rotação, quaternions e operações vetoriais associadas ao movimento relativo.

A implementação modular e a existência de testes automatizados representam também aspetos relevantes da solução desenvolvida, permitindo estabelecer procedimentos de verificação das operações matemáticas.

Em síntese, o projeto evoluiu de uma proposta de visualização de dados para uma ferramenta integrada de reprodução e análise de missões robóticas, mantendo uma relação direta entre os conceitos matemáticos estudados e a respetiva aplicação computacional.

## 2. Descrição do código implementado



### 2.1. Arquitetura geral da aplicação

O projeto foi desenvolvido em Python, recorrendo a uma arquitetura modular que permite separar as diferentes etapas de processamento, análise e visualização dos dados.

Esta organização teve como principal objetivo facilitar a manutenção do código, evitar a repetição de operações e permitir que as funcionalidades matemáticas fossem implementadas e testadas independentemente da interface gráfica.

A estrutura principal do projeto encontra-se organizada da seguinte forma:

```text
TOMSA-main/
│
├── app.py
├── config/
│   ├── config.yaml
│   └── esquema.py
│
├── io_dados/
│   ├── leitor.py
│   └── exportacao.py
│
├── geo/
│   └── transformacoes.py
│
├── analysis/
│   ├── reamostragem.py
│   ├── risco.py
│   ├── anomalias.py
│   ├── resumo.py
│   └── validacao.py
│
├── ui/
│   ├── layout.py
│   ├── callbacks.py
│   ├── mapa.py
│   ├── graficos.py
│   └── estilo.py
│
├── ai/
│   ├── ia_local.py
│   └── voz.py
│
└── tests/
    ├── test_analysis.py
    ├── test_config.py
    ├── test_dados.py
    ├── test_geodesia.py
    └── test_ia_local.py
```

O ficheiro `app.py` constitui o ponto de entrada da aplicação. É responsável por carregar a configuração, iniciar o processamento dos dados, preparar os indicadores da missão e criar o servidor Dash.

Os restantes módulos possuem responsabilidades específicas, permitindo estabelecer um fluxo de processamento desde os ficheiros CSV até à apresentação dos resultados no dashboard.

De forma simplificada, o funcionamento do sistema pode ser representado pela seguinte sequência:

**Ficheiros CSV e configuração YAML → Leitura e limpeza → Transformações geodésicas → Sincronização temporal → Análise dos dados → Visualização e exportação.**

Esta separação permite que as operações matemáticas mais exigentes sejam executadas durante a preparação da missão, enquanto a interface utiliza os dados previamente processados.

### 2.2. Leitura, configuração e preparação dos dados

A configuração da aplicação é realizada através do ficheiro `config/config.yaml`, permitindo definir os robôs envolvidos na missão, os respetivos ficheiros CSV, as cores de representação e a referência geográfica utilizada nas transformações de coordenadas.

O ficheiro permite também configurar parâmetros relacionados com a sincronização temporal, a identificação de anomalias, a exportação de dados e a componente de inteligência artificial.

A leitura e validação da configuração são realizadas no módulo `config/esquema.py`, recorrendo às bibliotecas PyYAML e Pydantic. A primeira permite interpretar o conteúdo do ficheiro YAML, enquanto a segunda verifica a estrutura e os valores fornecidos, reduzindo a possibilidade de erros de configuração.

O processamento dos registos dos robôs é realizado no módulo `io_dados/leitor.py`, que utiliza a biblioteca pandas para ler e organizar os dados.

Durante esta etapa são executadas operações de limpeza, nomeadamente a remoção de registos com valores em falta nas colunas essenciais, a eliminação de instantes temporais duplicados e a ordenação cronológica das observações.

Os dados incluem informação relativa à posição geográfica, orientação e velocidade dos veículos, proveniente dos registos ROS disponibilizados para a realização do trabalho.

Uma decisão importante foi a utilização de uma **origem temporal comum aos três robôs**, calculada a partir do primeiro registo disponível da missão. Esta abordagem permite que o mesmo valor de tempo relativo corresponda ao mesmo instante real, independentemente do veículo considerado.

Após a limpeza, os dados são convertidos para uma estrutura comum, contendo as posições geográficas, coordenadas NED, quaternions de orientação, *heading* e velocidade.

### 2.3. Transformações de coordenadas e orientação

O módulo `geo/transformacoes.py` concentra as principais operações matemáticas de representação espacial utilizadas no projeto.

A implementação destas transformações constitui uma aplicação direta dos conceitos de transformações lineares, matrizes de rotação e mudança de referencial estudados na unidade curricular.

#### 2.3.1. Conversão de coordenadas geográficas para ECEF

Os ficheiros de entrada disponibilizam a posição dos robôs através de latitude, longitude e altitude, habitualmente representadas por LLA (*Latitude, Longitude, Altitude*).

Contudo, estas coordenadas não são adequadas para calcular diretamente distâncias e deslocamentos através de operações vetoriais cartesianas.

Por esse motivo, foi implementada a função `lla2ecef()`, responsável pela conversão das coordenadas geográficas para o sistema ECEF (*Earth-Centered, Earth-Fixed*).

Esta transformação utiliza o modelo elipsoidal WGS84, permitindo representar cada posição através de três coordenadas cartesianas, medidas em metros.

Considerando a latitude $\varphi$, a longitude $\lambda$, a altitude $h$, o semieixo maior $a$ e a excentricidade ao quadrado $e^2$, define-se:

$$
N(\varphi)=\frac{a}{\sqrt{1-e^2\sin^2\varphi}}
$$

As coordenadas ECEF são obtidas através de:

$$
x=(N(\varphi)+h)\cos\varphi\cos\lambda
$$

$$
y=(N(\varphi)+h)\cos\varphi\sin\lambda
$$

$$
z=(N(\varphi)(1-e^2)+h)\sin\varphi
$$

Estas expressões foram implementadas com recurso à biblioteca NumPy, permitindo processar vetorialmente as posições dos robôs.

#### 2.3.2. Transformação para o referencial NED

Após a conversão para ECEF, as posições são transformadas para o referencial local NED (*North-East-Down*).

Este sistema possui três eixos: Norte ($N$), Este ($E$) e Baixo ($D$), sendo particularmente adequado à representação do movimento de veículos marinhos.

A origem do referencial é definida pela referência geográfica disponibilizada no ficheiro de configuração.

A transformação é realizada pela função `lla2ned()`, utilizando a matriz de rotação definida em `matriz_ned()`.

Considerando $\mathbf{p}_{ECEF}$ a posição de um robô e $\mathbf{p}_0$ a posição ECEF da origem, obtém-se:

$$
\mathbf{p}_{NED}
=
\mathbf{R}_{NED}
\left(
\mathbf{p}_{ECEF}-\mathbf{p}_0
\right)
$$

A matriz $\mathbf{R}_{NED}$ permite expressar o vetor de deslocamento no referencial local.

Uma propriedade importante desta transformação é a ortogonalidade da matriz de rotação:

$$
\mathbf{R}_{NED}^{T}\mathbf{R}_{NED}=\mathbf{I}
$$

Consequentemente, a componente de rotação inversa é obtida através da transposta da matriz, propriedade aplicada na função `ned2lla()` juntamente com a reposição da origem ECEF.

A implementação inclui também a conversão inversa de coordenadas ECEF para LLA, através da função `ecef2lla()`.

#### 2.3.3. Representação da orientação através de quaternions

A orientação dos robôs é disponibilizada nos registos através de quaternions, representados por quatro componentes:

$$
q=(q_x,q_y,q_z,q_w)
$$

Esta representação permite descrever rotações tridimensionais sem depender exclusivamente de ângulos de Euler.

No projeto, a biblioteca SciPy é utilizada para interpretar os quaternions e obter informações de orientação, nomeadamente o *heading* dos veículos.

A função `heading_de_quaternion()` realiza esta conversão, considerando a convenção de orientação definida no ficheiro de configuração.

Esta operação é particularmente importante para a vista local, uma vez que permite determinar a direção para a qual o robô de referência se encontra orientado.

#### 2.3.4. Transformação para o referencial do corpo

Para representar a posição de um robô relativamente a outro, não é suficiente conhecer as respetivas coordenadas globais. É também necessário considerar a orientação do veículo escolhido como referência.

O primeiro passo consiste em calcular a diferença entre as posições dos dois veículos no referencial NED:

$$
\Delta\mathbf{p}
=
\mathbf{p}_{vizinho}-\mathbf{p}_{referência}
$$

Para uma representação no plano horizontal, a função `ned_para_corpo()` utiliza o *heading* $\psi$ do robô de referência para transformar o deslocamento relativo:

$$
x_b=\Delta N\cos\psi+\Delta E\sin\psi
$$

$$
y_b=-\Delta N\sin\psi+\Delta E\cos\psi
$$

Nesta representação, $x_b$ corresponde ao deslocamento para a frente do robô e $y_b$ ao deslocamento para a direita.

O projeto inclui ainda a função `ned_para_corpo_completo()`, que considera a orientação tridimensional definida pelo quaternion.

Admitindo que a matriz $\mathbf{R}$ representa a orientação do corpo relativamente ao referencial NED, a transformação inversa é dada por:

$$
\Delta\mathbf{p}_{corpo}
=
\mathbf{R}^{T}\Delta\mathbf{p}_{NED}
$$

Esta operação permite representar os restantes veículos segundo os eixos associados ao robô selecionado, considerando as rotações de *roll*, *pitch* e *yaw*. Na visualização, o eixo vertical do corpo é apresentado como «Cima», pelo que o sinal da componente vertical resultante é invertido relativamente à convenção NED («Baixo»).

O módulo implementa também a função `ned_para_aer()`, que permite obter azimute, elevação e distância a partir de um deslocamento expresso em NED.

### 2.4. Sincronização e interpolação temporal

Os registos de telemetria dos diferentes robôs não apresentam necessariamente medições nos mesmos instantes.

Esta diferença constitui um problema para a análise relativa, uma vez que o cálculo de distâncias e posições entre veículos exige a comparação de estados correspondentes ao mesmo instante temporal.

Para resolver esta situação, foi desenvolvido o módulo `analysis/reamostragem.py`.

A função `grelha_comum()` cria uma sequência de instantes partilhada pelos três robôs. Na configuração fornecida com o projeto, o intervalo entre amostras é de um segundo.

Posteriormente, a função `reamostrar_robo()` estima os valores necessários nos instantes definidos pela grelha.

Para a posição e outras grandezas escalares, é utilizada interpolação linear. Considerando duas amostras nos instantes $t_0$ e $t_1$, a estimativa de uma grandeza $x$ num instante intermédio $t$ pode ser expressa por:

$$
x(t)=x(t_0)+
\frac{t-t_0}{t_1-t_0}
\left[x(t_1)-x(t_0)\right]
$$

No caso da orientação, a interpolação linear dos ângulos de Euler pode originar comportamentos incorretos, particularmente quando existe uma passagem entre ângulos próximos de $360^\circ$ e $0^\circ$.

Por esse motivo, foi utilizada a interpolação esférica de quaternions, conhecida como **SLERP (*Spherical Linear Interpolation*)**, disponibilizada pela biblioteca SciPy.

Esta abordagem permite interpolar a orientação respeitando a geometria das rotações tridimensionais.

Foi ainda implementado um mecanismo para impedir a interpolação através de falhas temporais demasiado longas. Na configuração utilizada, intervalos superiores a cinco segundos não são preenchidos artificialmente.

Assim, o sistema evita apresentar posições e orientações estimadas em períodos nos quais não existe informação suficiente.

### 2.5. Análise do movimento e proximidade entre veículos

Para além da representação das trajetórias, foram implementadas funcionalidades de análise do movimento dos robôs.

Estas operações encontram-se distribuídas pelos módulos `analysis/risco.py`, `analysis/resumo.py`, `analysis/anomalias.py` e `analysis/validacao.py`.

#### 2.5.1. Distâncias entre robôs

A posição de cada veículo é expressa no referencial NED, permitindo calcular a distância tridimensional entre dois robôs através da norma euclidiana:

$$
d_{AB}=
\sqrt{
(N_B-N_A)^2+
(E_B-E_A)^2+
(D_B-D_A)^2
}
$$

A função `distancia_minima_pares()` determina a menor distância observada entre cada par de veículos ao longo da missão, considerando os instantes em que ambos possuem dados válidos.

A função `alertas_proximidade()` identifica períodos em que a separação entre dois robôs é inferior ao limite definido. Na implementação, o valor utilizado para esta análise é de 20 metros.

#### 2.5.2. Estimativa de maior aproximação — DCPA e TCPA

O módulo `analysis/risco.py` implementa o cálculo da distância e do tempo até ao ponto de maior aproximação entre dois veículos, através dos indicadores DCPA (*Distance at Closest Point of Approach*) e TCPA (*Time to Closest Point of Approach*).

Ao contrário do cálculo das distâncias mínimas observadas, que considera as três componentes do referencial NED, esta análise utiliza apenas as componentes horizontais Norte e Este.

Assim, a posição de cada veículo é representada por:

$$
\mathbf{p}=
\begin{bmatrix}
N\\
E
\end{bmatrix}
$$

A velocidade horizontal é estimada a partir da velocidade escalar $v$ e do *heading* $\psi$:

$$
\mathbf{v}=
\begin{bmatrix}
v\cos\psi\\
v\sin\psi
\end{bmatrix}
$$

Esta aproximação assume que a direção do movimento coincide com a orientação horizontal do veículo.

Para dois robôs, A e B, definem-se os vetores relativos:

$$
\mathbf{p}_{rel}=\mathbf{p}_B-\mathbf{p}_A
$$

$$
\mathbf{v}_{rel}=\mathbf{v}_B-\mathbf{v}_A
$$

Assumindo velocidades constantes, o tempo correspondente ao ponto de maior aproximação é calculado através de:

$$
TCPA=
-\frac{\mathbf{p}_{rel}\cdot\mathbf{v}_{rel}}
{\|\mathbf{v}_{rel}\|^2}
$$

Quando a velocidade relativa não é praticamente nula, a distância horizontal prevista nesse instante é:

$$
DCPA=
\left\|
\mathbf{p}_{rel}+TCPA\,\mathbf{v}_{rel}
\right\|
$$

A implementação contempla situações particulares. Quando a velocidade relativa é aproximadamente nula, a distância atual é utilizada como valor de referência. Quando o TCPA calculado é negativo, considera-se que os veículos se encontram em afastamento, sendo devolvida a distância atual como indicador de separação.

É importante distinguir estes resultados das distâncias mínimas calculadas a partir da missão. A distância mínima observada considera a separação tridimensional entre os veículos, enquanto a DCPA representa uma estimativa de aproximação horizontal, baseada nas posições e velocidades relativas num determinado instante.

Além disso, a estimativa da velocidade horizontal a partir do *heading* pode apresentar limitações em veículos cujo deslocamento não acompanha a orientação, como pode acontecer em situações de deriva ou movimento lateral.

Esta limitação é particularmente relevante para o `robo_7`, cujos registos apresentam uma diferença significativa entre a orientação medida e a direção do deslocamento. Por esse motivo, os valores de DCPA/TCPA associados a esse veículo devem ser interpretados como indicadores aproximados e não como previsões exatas do movimento futuro.

#### 2.5.3. Deteção de anomalias e validação de velocidades

O módulo `analysis/anomalias.py` permite identificar irregularidades nos registos, incluindo saltos de posição, falhas temporais e acelerações anómalas.

Estas verificações permitem distinguir acontecimentos associados ao movimento dos veículos de possíveis problemas na qualidade da telemetria.

Adicionalmente, o módulo `analysis/validacao.py` compara a velocidade registada nos dados com uma estimativa calculada a partir da evolução das posições.

Esta comparação constitui uma forma de avaliar a coerência entre diferentes grandezas medidas ou calculadas durante a missão.

### 2.6. Interface gráfica e reprodução da missão

A interface foi desenvolvida através das bibliotecas Dash e Plotly, permitindo apresentar os resultados num dashboard web interativo.

O módulo `ui/layout.py` define a organização visual dos diferentes componentes, enquanto `ui/callbacks.py` estabelece as interações entre os controlos e os gráficos.

A representação geográfica é implementada em `ui/mapa.py`, através de um mapa interativo que apresenta as trajetórias dos três robôs, a origem do referencial NED e a posição atual de cada veículo.

A aplicação utiliza OpenStreetMap como mapa de fundo e permite acrescentar uma camada de informação marítima do OpenSeaMap. Quando não existe acesso à internet, o sistema pode apresentar as trajetórias sobre um fundo sem mosaicos geográficos.

Para reduzir o volume de informação transmitida durante a atualização dos gráficos, a aplicação utiliza o mecanismo `Patch()` do Dash em componentes cuja atualização é parcial.

Além das vistas global e local, o dashboard disponibiliza gráficos de evolução da velocidade, profundidade e distância entre robôs, bem como uma comparação entre velocidades medidas e estimadas.

### 3.5. Análise das distâncias e do movimento relativo

Para além da visualização das trajetórias, a aplicação disponibiliza ferramentas de análise quantitativa do movimento dos robôs.

Uma das funcionalidades implementadas permite determinar a distância tridimensional entre cada par de veículos, utilizando as respetivas posições no referencial NED.

A partir destas distâncias, são identificados os instantes de maior aproximação observados durante a missão.

A aplicação apresenta igualmente a evolução temporal das distâncias entre veículos, permitindo identificar períodos em que estes se encontram mais próximos ou mais afastados.

No módulo de análise, foi definido um limiar de proximidade de 20 metros, utilizado para identificar episódios em que a separação entre dois veículos é inferior a esse valor.

A deteção destes episódios permite caracterizar situações relevantes durante a missão, nomeadamente a aproximação entre diferentes veículos.

Adicionalmente, foram implementados os indicadores DCPA (*Distance at Closest Point of Approach*) e TCPA (*Time to Closest Point of Approach*).

Estes indicadores permitem estimar a distância mínima horizontal futura e o tempo necessário para a atingir, considerando as posições e velocidades relativas no plano Norte–Este e assumindo velocidades constantes.

Os resultados são utilizados para classificar as situações de aproximação segundo diferentes níveis de risco. A estimativa utiliza a direção do *heading* como aproximação à direção da velocidade horizontal, hipótese que pode não se verificar em movimentos laterais ou de deriva.

É importante distinguir as duas análises realizadas. As distâncias mínimas observadas são calculadas no espaço tridimensional, considerando as componentes Norte, Este e Baixo (NED) das posições dos veículos. Por outro lado, os indicadores DCPA e TCPA utilizam apenas as componentes horizontais Norte e Este, permitindo estimar a maior aproximação no plano horizontal, assumindo velocidades constantes. Assim, enquanto a distância mínima observada representa a menor separação tridimensional registada durante a missão, os indicadores DCPA/TCPA fornecem estimativas de aproximação futura, não considerando diretamente a separação vertical entre os veículos.

#### 3.8.1. Validação das transformações geodésicas

A validação matemática do projeto incluiu testes destinados a verificar a consistência das transformações de coordenadas implementadas no módulo `geo/transformacoes.py`.

Um dos testes verifica que a origem geográfica definida para a missão corresponde ao ponto $(0,0,0)$ no referencial NED.

Foi também verificada a ortogonalidade da matriz de transformação entre os referenciais ECEF e NED, através da relação:

$$
\mathbf{R}_{NED}\mathbf{R}_{NED}^{T}=\mathbf{I}
$$

Esta propriedade é importante porque permite utilizar a transposta da matriz para realizar a transformação inversa.

Foram igualmente realizados testes de transformação inversa, nomeadamente LLA → NED → LLA e LLA → ECEF → LLA, para verificar a recuperação das coordenadas iniciais após sucessivas conversões.

A implementação própria das transformações geodésicas foi comparada com bibliotecas de referência:

- **`pyproj`:** validação da conversão de coordenadas geográficas para ECEF.
- **`pymap3d`:** validação da conversão de coordenadas geográficas para NED.

Estas comparações foram realizadas sobre um conjunto de pontos de teste definidos nas proximidades da origem geográfica da missão.

A validação foi realizada através da função `numpy.testing.assert_allclose()`, que permite comparar os resultados da implementação própria com os valores obtidos através das bibliotecas de referência, considerando tolerâncias numéricas.

Nos testes das coordenadas cartesianas, foi definido o parâmetro de tolerância absoluta `atol=1e-3`, correspondente a 0,001 metros. No entanto, a função considera também uma tolerância relativa, aplicada por omissão. Assim, o critério de aceitação combina ambas as tolerâncias:

$$
|x-x_{\mathrm{ref}}|
\leq
atol+rtol\cdot|x_{\mathrm{ref}}|
$$

Deste modo, a aprovação dos testes não significa necessariamente que o erro absoluto máximo seja de 1 mm, mas sim que os resultados satisfazem os critérios de concordância numérica estabelecidos.

Adicionalmente, foram realizados testes às operações de orientação, incluindo a conversão de quaternions para *heading*, a transformação para o referencial do corpo e a conversão de coordenadas NED para azimute, elevação e distância.

A validação da coerência entre o *heading* e a direção do movimento foi realizada para os robôs `robo_1` e `robo_4`, utilizando as posições NED dos registos reais.

Os testes automatizados executados apresentaram resultados positivos nos casos considerados, constituindo uma evidência da consistência das transformações implementadas.

Importa salientar que estes resultados demonstram a concordância nos pontos e nas condições testadas, não representando uma garantia de erro máximo para todas as coordenadas geográficas possíveis.

Os resultados mostram que o par constituído pelos robôs `robo_1` e `robo_4` apresentou a menor separação durante a missão, atingindo aproximadamente **27,7 metros**, no instante 3249 segundos.

Para os restantes pares, as distâncias mínimas foram significativamente superiores, correspondendo a 175,0 metros entre `robo_1` e `robo_7` e a 206,7 metros entre `robo_4` e `robo_7`.

A função `alertas_proximidade()` foi também utilizada para identificar episódios em que a distância tridimensional entre dois veículos fosse inferior ao limiar de 20 metros definido na aplicação.

Não foram identificados episódios abaixo desse limite, uma vez que todas as distâncias mínimas observadas foram superiores a 20 metros.

Estes resultados permitem caracterizar a separação espacial entre os veículos durante a missão e demonstram a aplicação das operações vetoriais no referencial NED.

#### 3.9.2. Validação das velocidades

A coerência das velocidades registadas foi avaliada através da comparação entre a velocidade disponibilizada na telemetria e uma estimativa obtida a partir da evolução temporal das posições NED.

Esta análise foi realizada através da função `comparar_velocidades()`, implementada no módulo `analysis/validacao.py`.

A estimativa da velocidade é calculada a partir das derivadas numéricas das componentes Norte, Este e Baixo, utilizando posteriormente a norma do vetor resultante:

$$
v_{\text{estimada}} =
\sqrt{
\left(\frac{dN}{dt}\right)^2 +
\left(\frac{dE}{dt}\right)^2 +
\left(\frac{dD}{dt}\right)^2
}
$$

Para reduzir o efeito do ruído das medições, a implementação aplica uma média móvel centrada de cinco amostras às velocidades comparadas.

A diferença absoluta entre as velocidades é utilizada para calcular o erro médio absoluto e o erro mediano absoluto de cada robô.

| Robô | Erro médio absoluto (m/s) | Erro mediano absoluto (m/s) |
|---|---:|---:|
| `robo_1` | 0,014 | 0,005 |
| `robo_4` | 0,019 | 0,005 |
| `robo_7` | 0,052 | 0,025 |

Os resultados mostram que os robôs `robo_1` e `robo_4` apresentam erros médios absolutos inferiores a 0,02 m/s, enquanto o `robo_7` apresenta um erro médio absoluto de 0,052 m/s.

Embora os valores obtidos indiquem uma concordância relativamente próxima entre as velocidades registadas e as velocidades estimadas, verificam-se diferenças entre os três veículos.

Estas diferenças podem estar relacionadas com as características do movimento, com o ruído das medições e com os efeitos do processamento temporal.

Assim, a comparação constitui um indicador da coerência dos dados de telemetria, sem representar uma garantia de ausência de erros nas medições.

#### 3.9.3. Análise da qualidade dos dados

A análise da qualidade dos registos foi realizada através das funções implementadas no módulo `analysis/anomalias.py`, utilizando os limiares definidos no ficheiro `config.yaml`.

Foram considerados três tipos de acontecimentos: saltos de posição, falhas temporais e acelerações anómalas.

| Indicador | Limiar utilizado | Ocorrências identificadas |
|---|---|---:|
| Saltos de posição | Velocidade aparente superior a 10 m/s | 3 |
| Falhas temporais | Intervalos entre amostras superiores a 2 s | 0 |
| Acelerações anómalas | Módulo da aceleração superior a 1 m/s² | 3 |

A deteção de saltos de posição identificou **três ocorrências** nas quais a velocidade aparente, calculada a partir da diferença entre posições consecutivas, ultrapassou o limiar estabelecido.

Relativamente às falhas temporais, não foram identificados intervalos superiores a dois segundos entre amostras consecutivas dos registos analisados.

A análise das acelerações identificou igualmente três ocorrências acima do limiar configurado, calculadas a partir da variação temporal da velocidade escalar registada.

Estes acontecimentos representam possíveis anomalias, não sendo possível concluir, apenas através dos critérios utilizados, que correspondem necessariamente a erros dos sensores.

A identificação destas situações permite complementar a análise das trajetórias e constitui um procedimento relevante para avaliar a qualidade dos dados utilizados.

### 3.10. Síntese dos resultados

Os resultados obtidos demonstram que a aplicação permite processar, sincronizar e analisar os registos de telemetria dos três veículos utilizados na missão.

A transformação das coordenadas geográficas para o referencial NED possibilitou o cálculo das posições relativas e das distâncias tridimensionais entre veículos.

A análise das distâncias permitiu identificar que a menor separação tridimensional observada ocorreu entre `robo_1` e `robo_4`, com aproximadamente 27,7 metros, não tendo sido registados episódios de proximidade inferiores ao limiar de 20 metros definido para essa análise.

A comparação das velocidades registadas com as estimadas a partir das posições apresentou erros médios absolutos relativamente reduzidos, embora com diferenças entre os três veículos.

Adicionalmente, os mecanismos de análise da qualidade dos dados permitiram identificar possíveis anomalias de posição e aceleração.

A validação do sistema foi complementada pelos testes automatizados, cuja execução resultou em **45 testes aprovados e um teste ignorado**, devido à ausência de uma dependência opcional.

Em conjunto, estes resultados evidenciam a aplicação prática de conceitos matemáticos de TOMSA, nomeadamente transformações de coordenadas, operações vetoriais, matrizes de rotação e análise do movimento relativo de sistemas robóticos autónomos.

## 4. Análise crítica do trabalho da equipa

### 4.1. Organização do desenvolvimento

Na primeira entrega, a equipa previu uma distribuição das tarefas entre o processamento dos dados e a implementação matemática, por um lado, e o desenvolvimento da interface e das funcionalidades interativas, por outro. Esta organização está refletida na separação do projeto em módulos de configuração, geodesia, análise, entrada/saída de dados e interface.

A integração destes módulos exigiu definir estruturas de dados partilhadas e garantir que os resultados matemáticos utilizados nos gráficos correspondiam aos mesmos instantes temporais. A arquitetura final separa o pré-processamento da atualização interativa dos gráficos, evitando repetir as transformações geodésicas a cada atualização do *replay*.

### 4.2. Decisões técnicas e limitações

Uma decisão relevante foi implementar as transformações de coordenadas com NumPy e verificar os seus resultados através de bibliotecas de referência. Também se optou pela utilização de quaternions e SLERP para tratar a orientação ao longo do tempo, em vez de interpolar diretamente ângulos de Euler.

A análise DCPA/TCPA utiliza um modelo horizontal de velocidade constante e uma estimativa da direção da velocidade a partir do *heading*. Esta hipótese não descreve necessariamente movimentos laterais ou de deriva e constitui uma limitação da interpretação dos indicadores, especialmente quando a orientação e o deslocamento não coincidem.

A execução dos testes no ambiente de desenvolvimento exigiu isolar os plugins do ROS, pois o carregamento automático desses plugins provocava um erro no `pytest`. Com a desativação desse carregamento, a execução registada terminou com 45 testes aprovados e um teste ignorado devido à ausência da dependência opcional `espeak-ng`.
